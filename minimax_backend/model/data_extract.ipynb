```python
from minimax_backend.prediction_all import make_smiles

remove_lst = []

for smiles in df['SMILES']:
    try:
        make_smiles(smiles)
    except:
        remove_lst.append(smiles)
```

```python
from rdkit import Chem
from rdkit.Chem.Draw import rdMolDraw2D
import base64, json

def smiles_to_svg_base64(smiles: str, size=(300, 300)) -> str:
    mol = Chem.MolFromSmiles(smiles)
    drawer = rdMolDraw2D.MolDraw2DSVG(size[0], size[1])
    rdMolDraw2D.PrepareAndDrawMolecule(drawer, mol)
    drawer.FinishDrawing()
    svg = drawer.GetDrawingText()
    # Base64 인코딩
    return base64.b64encode(svg.encode("utf-8")).decode("utf-8")
```

```python
from chembl_webresource_client.new_client import new_client

molecule = new_client.molecule
df_unique[['d_chembl_id','d_name','d_canosmiles','d_formula','d_type','d_image_base64']] = None

for i in range(len(df_unique)):
    drug_name = df_unique.loc[i, '치료제']
    res = molecule.filter(pref_name__icontains=drug_name)
    
    if res:
        mol = res[0]
        df_unique.loc[i, 'd_chembl_id'] = mol.get("molecule_chembl_id")
        df_unique.loc[i, 'd_name'] = mol.get("pref_name")
        df_unique.loc[i, 'd_canosmiles'] = mol["molecule_structures"]["canonical_smiles"]
        df_unique.loc[i, 'd_formula'] = mol.get("molecule_properties", {}).get("full_molformula")
        df_unique.loc[i, 'd_type'] = mol.get("molecule_type")
```

```python
def categorize(detail):
    if detail in ['위암', '간암', '자궁경부암', '대장암', '유방암']:
        return '암 치료제'
    elif detail in ['에이즈 (HIV)', '코로나19', '인플루엔자', 'B형 간염', 'C형 간염']:
        return '항바이러스제'
    elif detail in ['뇌전증', '치매', '뇌졸중', '파킨슨병']:
        return '신경 질환'
    else:
        return '대사 질환'

df_unique['d_category'] = df_unique['d_detail_category'].apply(categorize)
```
