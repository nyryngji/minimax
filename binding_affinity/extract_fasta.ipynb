```python
import os
from Bio import SeqIO

fastas = []

# 폴더 내의 fasta 파일 중 앞의 2개만 처리
for fasta_file in os.listdir(r'D:\minimax\binding_affinity\질환관련 fasta'):
    disease_name = fasta_file.split('.')[0]
    fasta_file_name = os.path.join(r'D:\minimax\binding_affinity\질환관련 fasta', fasta_file)
    sequences = {}
    for record in SeqIO.parse(fasta_file_name, "fasta"):
        protein_id = record.id        # 단백질 ID (예: P00533)
        sequence = str(record.seq)    # 아미노산 서열
        sequences['타겟 질환'] = disease_name
        sequences['fasta'] = sequence
    fastas.append(sequences)   # print는 루프 밖에서 한 번만 출력
```

```python
import pandas as pd 
fasta_info = pd.DataFrame(fastas)
fasta_info.head()
```
